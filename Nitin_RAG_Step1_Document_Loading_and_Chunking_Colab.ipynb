# RAG Step 1 — Document Loading and Chunking

This notebook loads a sample document and splits it into smaller chunks for a Retrieval-Augmented Generation (RAG) pipeline.

## 1. Install and import libraries

```python
!pip -q install langchain-text-splitters

from langchain_text_splitters import RecursiveCharacterTextSplitter
print("Libraries installed and imported successfully.")
```

## 2. Create the source document

```python
document = (
    "Retrieval-Augmented Generation (RAG) combines information retrieval with language generation.\n"
    "A RAG system first stores documents in a searchable form.\n"
    "When a user asks a question, relevant pieces of the documents are retrieved.\n"
    "The retrieved context is then provided to a language model to generate a grounded answer.\n"
    "Document chunking is important because very large documents are difficult to retrieve efficiently.\n"
    "Good chunks should preserve enough context while remaining small enough for accurate retrieval."
)
print(document)
```

## 3. Split the document into chunks

```python
splitter = RecursiveCharacterTextSplitter(chunk_size=180, chunk_overlap=30)
chunks = splitter.split_text(document)

print("Number of chunks:", len(chunks))
for i, chunk in enumerate(chunks, 1):
    print(f"\n--- Chunk {i} ---")
    print(chunk)
```

## 4. Inspect chunk sizes

```python
for i, chunk in enumerate(chunks, 1):
    print(f"Chunk {i}: {len(chunk)} characters")
```

## 5. Summary

```python
print("Document loading: completed")
print("Chunking: completed")
print("Chunks created:", len(chunks))
print("These chunks are ready for the next RAG step: vector database storage.")
```
