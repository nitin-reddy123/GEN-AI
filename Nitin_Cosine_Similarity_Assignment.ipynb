## Example 1 — Calculating Cosine Similarity Mathematically

**Goal:** Calculate the similarity between two numeric vectors in three ways: manually, with NumPy, and with scikit-learn.

```python
vector_A = [3, 1, 2]
vector_B = [2, 4, 1]

print("Vector A:", vector_A)
print("Vector B:", vector_B)
```

```python
def dot_product(A, B):
    total = 0
    for a, b in zip(A, B):
        total += a * b
    return total

dp = dot_product(vector_A, vector_B)
print("Dot Product (A . B):", dp)
```

```python
import math

def magnitude(A):
    total = 0
    for a in A:
        total += a ** 2
    return math.sqrt(total)

mag_A = magnitude(vector_A)
mag_B = magnitude(vector_B)

print("Magnitude of A:", round(mag_A, 4))
print("Magnitude of B:", round(mag_B, 4))
```

Cosine similarity formula:

\[
\text{cosine similarity}(A,B)=\frac{A\cdot B}{\|A\|\,\|B\|}
\]

The dot product is 12. The magnitudes are approximately 3.7417 and 4.5826.

```python
def cosine_similarity_manual(A, B):
    dp = dot_product(A, B)
    mag_A = magnitude(A)
    mag_B = magnitude(B)
    if mag_A == 0 or mag_B == 0:
        raise ValueError("Cosine similarity is undefined for a zero vector.")
    return dp / (mag_A * mag_B)

result_manual = cosine_similarity_manual(vector_A, vector_B)
print("Cosine Similarity (Manual Calculation):", round(result_manual, 4))
```

```python
import numpy as np

A = np.array(vector_A)
B = np.array(vector_B)

result_numpy = np.dot(A, B) / (np.linalg.norm(A) * np.linalg.norm(B))
print("Cosine Similarity (NumPy):", round(result_numpy, 4))
```

```python
from sklearn.metrics.pairwise import cosine_similarity

A_2d = np.array(vector_A).reshape(1, -1)
B_2d = np.array(vector_B).reshape(1, -1)

result_sklearn = cosine_similarity(A_2d, B_2d)[0][0]
print("Cosine Similarity (scikit-learn):", round(result_sklearn, 4))

print("\nManual      :", round(result_manual, 6))
print("NumPy       :", round(result_numpy, 6))
print("scikit-learn:", round(result_sklearn, 6))
assert round(result_manual, 4) == round(result_numpy, 4) == round(result_sklearn, 4)
print("All three methods match!")
```

**Expected result:** cosine similarity is approximately **0.6999** for these vectors. The three methods should agree up to rounding.

**Conclusion:** Cosine similarity compares vector direction. It is widely used to compare text embeddings in semantic search.
