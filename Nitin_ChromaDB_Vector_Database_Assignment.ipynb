## Example 3 — ChromaDB Vector Database with Books

**Goal:** Store book-related documents in ChromaDB, retrieve relevant records, filter by metadata, update/delete records, and try a Hugging Face embedding function.

```python
!pip -q install chromadb sentence-transformers
```

```python
import chromadb

client = chromadb.Client()
print("ChromaDB client created successfully!")

collection = client.create_collection(name="assignment_book_collection")
```

```python
collection.add(
    documents=[
        "Python is a popular language for programming and data science.",
        "The Great Gatsby is a famous American novel.",
        "Machine learning allows computers to learn patterns from data.",
        "Harry Potter is a popular fantasy book series.",
        "Deep learning uses neural networks to solve complex problems.",
        "Sherlock Holmes is a famous detective character created by Arthur Conan Doyle."
    ],
    ids=["book1", "book2", "book3", "book4", "book5", "book6"]
)

print("Documents added! Total documents:", collection.count())
```

```python
results = collection.query(
    query_texts=["Tell me about programming and machine learning"],
    n_results=2
)

for doc, distance in zip(results["documents"][0], results["distances"][0]):
    print(f"Distance: {distance:.4f} | {doc}")
```

```python
results = collection.query(
    query_texts=["Tell me about famous detective stories"],
    n_results=2
)

for doc, distance in zip(results["documents"][0], results["distances"][0]):
    print(f"Distance: {distance:.4f} | {doc}")
```

```python
collection.add(
    documents=[
        "The Hobbit is a fantasy adventure novel by J.R.R. Tolkien.",
        "Atomic Habits explains practical methods for building better habits."
    ],
    metadatas=[
        {"category": "fiction"},
        {"category": "self-help"}
    ],
    ids=["book7", "book8"]
)

results = collection.query(
    query_texts=["Tell me about fantasy adventures"],
    n_results=2,
    where={"category": "fiction"}
)

for doc, distance in zip(results["documents"][0], results["distances"][0]):
    print(f"Distance: {distance:.4f} | {doc}")
```

```python
from chromadb.utils import embedding_functions

hf_embedder = embedding_functions.SentenceTransformerEmbeddingFunction(
    model_name="all-MiniLM-L6-v2"
)

custom_collection = client.create_collection(
    name="assignment_custom_book_embeddings",
    embedding_function=hf_embedder
)

custom_collection.add(
    documents=[
        "I enjoy reading science fiction novels.",
        "Fantasy books contain magical worlds and adventures.",
        "Cooking books provide recipes and food preparation techniques."
    ],
    ids=["b1", "b2", "b3"]
)

results = custom_collection.query(
    query_texts=["books about magical adventures"],
    n_results=2
)

for doc, distance in zip(results["documents"][0], results["distances"][0]):
    print(f"Distance: {distance:.4f} | {doc}")
```

```python
collection.update(
    ids=["book4"],
    documents=[
        "Harry Potter is a popular fantasy book series about a young wizard and his adventures."
    ]
)
print("Updated book4")

collection.delete(ids=["book2"])
print("Deleted book2")
print("Remaining documents:", collection.count())
```

**Conclusion:** A vector database stores documents and retrieves semantically similar content. Metadata filters narrow results, while update and delete operations maintain the collection. Distances are database-specific distance values; generally, a smaller distance means a closer match.
